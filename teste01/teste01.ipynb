### 7. Corrigir — função que calcula a média [20 pontos]

A função media devia devolver a média de uma lista de notas, mas o código tem 3 erros. Executa, lê a última linha de cada mensagem de erro e corrige um erro de cada vez.

Em cada linha corrigida, acrescenta no fim um comentário # a dizer o que estava errado. Depois de corrigido, o programa mostra:

```text
Média: 12.0
```

````python
def media(notas)
    total = 0
    for nota in notas:
        total += nota
    return total / len(nota)

resultado = media([10, 14, 12])
print("Média: " + resultado)
````
