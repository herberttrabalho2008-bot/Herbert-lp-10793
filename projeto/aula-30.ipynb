# Exercicio 2

```python
import pandas as pd

df = pd.read_csv("acidentes.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["feridos"])

a2024 = df[df["ano"] == 2024]
print(a2024.shape)
porto = df[df["distrito"] == "Porto"]
print(porto["acidentes"].sum())
# O shape do filtro vai ter menos linhas do que o shape do df original, porque o programa está a filtrar apenas os acidentes do ano de 2024.  
```

# Exercicio 3

```python
import pandas as pd
df = pd.read_csv("acidentes.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["feridos"])
distracao = df[df["causa"] == "Distração"]
dist2024 = distracao[distracao["ano"] == 2024]
print(dist2024["acidentes"].sum())
dist2023 = distracao[distracao["ano"] == 2023]
print(dist2023["acidentes"].sum())
# os resoltados de 2024 são maiores que os de 2023
```
