# Exercicio 1

```python
import pandas as pd

df = pd.read_csv("acidentes.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["feridos"])
por_mes = df.groupby("mes")["feridos"].sum()
print(por_mes)
print(por_mes.max())
# O resultado vai ter 12 linhas, uma para cada mês do ano.
```

# Exercicio 2

```python
import matplotlib.pyplot as plt
import pandas as pd
# usa o df limpo (as linhas de leitura e limpeza são as do exercício 1)
df = pd.read_csv("acidentes.csv")
por_causa = df.groupby("causa")["acidentes"].sum()
por_causa.plot(kind="bar", color="steelblue")
plt.title("Acidentes por causa")
plt.xlabel("Causa")
plt.ylabel("Acidentes")
plt.tight_layout()
plt.show()
print(por_causa)
# vai aparecer 4 barras, uma para cada causa de acidente
```

# Exercicio 3

```python
import matplotlib.pyplot as plt

# usa o df limpo do exercício 1
a2024 = df[df["ano"] == 2024]
por_mes = a2024.groupby("mes")["acidentes"].sum()
por_mes.plot(kind="line", marker="o")
plt.title("Acidentes por mês em 2024")
plt.xlabel("Mês")
plt.ylabel("Acidentes")
plt.show()
print(por_mes.max())
# O grafico vai ter 12 pontos, linhas é o melhor tipo de grafico para mostrar a evolução ao longo do tempo.
```

# Exercicio 4

```python
def variacao(antes, depois):
    return round((depois - antes) * 100 / antes, 1)

def test_variacao():
    assert variacao(100, 125) == 25.0
    assert variacao(200, 100) == -50.0
    assert variacao(80, 80) == 0.0
```
