## 2. Leitura e limpeza

```python
import pandas as pd

df = pd.read_csv("acidentes.csv")
df = df.drop_duplicates()
df = df.dropna()
# Eu fiz o dropna para remover as linhas sem valor geral sem ser algo especifico
# lê o ficheiro acidentes.csv
# tira as linhas repetidas e as linhas sem valor em feridos
```

## 3. Seleção

Filtra os dados (por exemplo, só um ano ou um distrito).

```python
import pandas as pd

df = pd.read_csv("acidentes.csv")

df = df.sort_values(by="feridos", ascending=False)
df = df.sort_values(by="mes", ascending=False)

df_2024 = df[df["ano"] == 2024]

# o teu filtro
```

## 4. Agrupamento

Agrupa com groupby e soma.

```python
import pandas as pd

df = pd.read_csv("acidentes.csv")
# Pergunta 1: quantos feridos houve em cada distrito?
feridos_por_distrito = df_2024.groupby("distrito")["feridos"].sum()
print(feridos_por_distrito)
# Mostra o total de feridos de cada distrito, do mais para o menos afetado.

# Pergunta 2: como evoluiu o número de feridos ao longo dos anos?
feridos_por_mes = df_2024.groupby("mes")["feridos"].sum()
print(feridos_por_mes)
# Mostra o total de feridos em cada mês, para ver se a tendência sobe ou desce.
```

## 5. Gráficos

Um gráfico de barras e um gráfico de linhas, com título e nome dos eixos.

```python
import matplotlib.pyplot as plt

# gráfico de barras
feridos_por_distrito.plot(kind="bar")
plt.title("Total de feridos por distrito em 2024")
plt.xlabel("Distrito")
plt.ylabel("Número de feridos")
plt.tight_layout()
plt.show()
```

```python
# gráfico de linhas
feridos_por_mes.plot(kind="line", marker="o")
plt.title("Evolução do total de feridos por mês em 2024")
plt.xlabel("Mês")
plt.ylabel("Número de feridos")
plt.grid()
plt.tight_layout()
plt.show()
```
