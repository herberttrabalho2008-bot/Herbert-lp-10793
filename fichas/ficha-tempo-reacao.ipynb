# Exercicio

```python
import pandas as pd
df = pd.read_csv("tempo_reacao.csv")
print(df.shape)
print(df.head())
print(df.isna().sum())
print(df.duplicated().sum())
# tem duas tempos em falta e 1 linha repetida.
```

# Exercicio 2

```python
df = df.drop_duplicates()
df = df.dropna(subset=["tempo_ms"])
df["tempo_ms"] = df["tempo_ms"].astype(int)
print(df.shape)
print(df.dtypes)
# passou de 25 linhas para 22.
```

# Exercicio 3

```python
media_mao = df.groupby("mao")["tempo_ms"].mean().round(1)
print(media_mao)
media_tentativa = df.groupby("tentativa")["tempo_ms"].mean().round(1)
print(media_tentativa)
# A mão da direita é mais rapida que a mão da esquerda.
```

# Exercicio 4

```python
import matplotlib.pyplot as plt
media_tentativa.plot(kind="line", marker="o")
plt.title("Tempo Médio por Tentativa")
plt.xlabel("tentativa")
plt.ylabel("tempo médio (ms)")
plt.show()
```

# Exercício 5

```python
media_mao.plot(kind="bar")
plt.title("Tempo medio por tentativa")
plt.xlabel("tentativa")
plt.ylabel("tempo médio (ms)")
plt.tight_layout()
plt.savefig("media_tentativa.png")
plt.show()
```

# Extenções 2

```python
import pandas as pd
import matplotlib.pyplot as plt
sond = pd.read_csv("sondagem.csv")
resultados = sond.groupby("resposta")["votos"].sum()
print(resultados)
resultados.plot(kind="bar")
plt.title("Resultados da Sondagem")
plt.xlabel("Resposta")
plt.ylabel("Número de Votos")
plt.tight_layout()
plt.savefig("sondagem.png")
plt.show()
```
