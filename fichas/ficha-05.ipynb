#Exercicio 1

```python
pontos = 10
#pontos tem o valor inicial de 10
pontos += 5
#pontos tem o valor 15
pontos *= 2
#pontos tem o valor 30
pontos -= 4
#pontos tem o valor 26
pontos //= 3
#pontos tem o valor 8
print(pontos)
```

#Exercicio 3

```python
cidade = "  lousada  "
print(len(cidade))
#vai dar 11 porque conta os espaços no início e no fim
cidade = cidade.strip()
#vai remover os espaços no início e no fim
print(len(cidade))
#vai dar 7 porque remove os espaços no início e no fim
print(cidade.upper())
#vai colocar a variavel cidade em maiúsculas "LOUSADA"
print(cidade.title())
#vai colocar a primeira letra em maiúscula "Lousada"
print(cidade.count("a"))
#vai contar quantas vezes a letra "a" aparece na variavel cidade, vai dar 2
print(cidade.replace("a", "@"))
#vai substituir todas as letras "a" por "@", vai dar "lous@d@"
```

#Exercicio 4

```python
saldo = 20.0
saldo -= 2.5    # paga uma refeição
saldo += 10     # carrega 10 €
saldo -= 2.5    # paga outra refeição
print(f"Saldo: {saldo:.2f} €")
```
