#Exercício 2: prevê e completa

```python
print(int("8") + 2) 
print(float("8") + 2) 
print(str(8) + "2") 
print(int(9.99)) 
print(int("9.99"))
```

```python
nome = input("Nome: ") 
idade = int(input("Idade: ")) 
altura = float(input("Altura em metros (ex.: 1.75): ")) 
idade_daqui_a_10 = idade + 10 
print(nome + " daqui a 10 anos terá " + str(idade_daqui_a_10) + " anos.") 
print(nome + " mede " + str(altura) + " m.")
```

Exercício 3: completa e prevê

```python
email = input("Email: ")
email = email.strip()     
# tira os espaços do início e do fim 
email = email.lower()     
# passa tudo para minúsculas 
print(email) 
print(len(email)) 
print(email.count(".")) 
print(email.replace("@", " (at) "))
```

#Exercício 4: completa a lista de preços

```python
produto1 = "Sopa" 
preco1 = 0.9 
produto2 = "Prato do dia" 
preco2 = 2.5 
print(f"{produto1:<15} {preco1:>5.2f} €") 
print(f"{produto2:<15} {preco2:>5.2f} €")
print("-" * 22) 
total = preco1 + preco2 
etiqueta = "Total" 
print(f"{etiqueta:<15}{total:>5.2f} €")
```
