```python
class Termometro:
    def __init__(self):
        self._graus = 20

    def ajustar(self, graus):
        if graus >= -10 and graus <= 40:
            self._graus = graus
        else:
            print("Fora do intervalo")

    def graus(self):
        return self._graus

t = Termometro()
# criar o objeto termometro
t.ajustar(25)
# comparando 25 e agora graus temo valor de 5
t.ajustar(90)
# mostra Fora de intervalo
print(t.graus())
# mostra 25
```

# Exercicio 2

```python
class Termometro:
    def __init__(self):
        self._graus = 20
    @property
    def graus(self):
        return self._graus

    @graus.setter
    def graus(self, valor):
        if valor >= -10 and valor <= 40:
            self._graus = valor
        else:
            print("Fora do intervalo")

    def graus(self, valor):
        if valor >= -10 and valor <= 40:
            self._graus = valor
       
t = Termometro()
t.graus = 25
t.graus = 90
print(t.graus)
```

# Exercicio 3

```python
class Livro:
    def __init__(self, titulo, autor, ano):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano

    def __str__(self):
        return f"{self.titulo}, de {self.autor} ({self.ano})"

l = Livro("Mensagem", "Fernando Pessoa", 1937)
print(l)
```

# Atividade

```python
class produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco
    @property
    def preco(self):
        return self.preco
    
    @preco.setter
    def preco(self, valor):
        if valor >= 0:
            self.preco = valor
        else:
            print("Preço inválido")

    def __str__(self):
        return f"{self.nome} - {self.preco:.2f} $"
    
    def __eq__(self, outro):
        return self.nome == outro.nome
    
p1 = produto("Caneta", 2.5)
p2 = produto("Caneta", 3.0)
p1.preco = -3
print(p1)
print(p1 == p2)
```
