# Exercicio 1

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos  WHERE nota >= 14")
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
#vai mostrar os alunos com nota maior ou igual a 14 que são Ana 16.0 e carla 18.0
```

# Exercicio 2

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.execute("SELECT nome, nota FROM alunos Where nota < ? Order by nota asc", (10,))
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
```

# Exercicio 3

```python
import sqlite3
ligacao = sqlite3.connect("escola.db")
cursor = ligacao.cursor()
cursor.executemany("INSERT INTO alunos (id, nome, nota) VALUES (?, ?, ?)", [(7, "herbert", 20.0),(8, "joao", 9.0),(9, "maria", 10.0)])
ligacao.commit()
cursor.execute("SELECT nome, nota FROM alunos")
for linha in cursor.fetchall():
    print(linha)
ligacao.close()
```
