## Exercício 3: completa a lista de preços

```python
precos = [1.5, 2.0, 0.8]
precos.append(3.2) # acrescenta 3.2 no fim
precos.remove(2.0) # tira o 2.0
precos.sort() # ordena por ordem crescente
print(precos)
total = sum(precos)
print(f"Total: {total:.2f} €")
# o total vai ser 5.50 € 
```

## Exercício 4: numera e soma

```python
precos = [1.5, 2.0, 0.8, 3.2]
precos.sort()
for i, preco in enumerate(precos, start=1):
    print(f"{i}. {preco:.2f} €")
total = sum(precos)
print(f"Total: {total:.2f} €")
```
