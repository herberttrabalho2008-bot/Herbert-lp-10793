## Exercício 2: inventário com get() e items()

```python
stock = {"teclado": 10, "rato": 25, "monitor": 4}
# 1. mostra a quantidade de "monitor" com get()
print(stock.get("monitor"))
# 2. mostra a quantidade de "webcam" com get(), com 0 por omissão
print(stock.get("webcam", 0))
# 3. percorre o stock todo e mostra cada produto no formato:
for produto, qtd in stock.items():
    print(f"   {produto}: {qtd} unidades")
#    teclado: 10 unidades
```
