# lista de tarefa guardada num ficheico

```python

def mostrar_tarefas():
    contador = 0
    try:
        with open("tarefas.txt", "r", encoding="utf-8") as f:
            for linha in f:
             contador += 1   
             print(f"{contador}: {linha.strip()}")
    except FileNotFoundError:
        print("O ficheiro tarefas.txt ainda não existe.")
    
    
def acrescentar_tarefa(tarefa):
    with open("tarefas.txt", "a", encoding="utf-8") as f:
        f.write(f"{tarefa}\n")
nova_tarefa = input("Digite uma nova tarefa: ")    
mostrar_tarefas()
acrescentar_tarefa(nova_tarefa)
mostrar_tarefas()
```
