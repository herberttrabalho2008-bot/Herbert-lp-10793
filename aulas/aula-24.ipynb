# Exercicio 1

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)
df["receita"] = df["preco"] * df["quantidade"]
receita = df.groupby("semana")["receita"].sum()
receita.plot(kind="line", marker="o")
plt.title("receita por semana")
plt.xlabel("Semana")
plt.ylabel("Euros")
plt.show()
print(receita.round(2))
```

# Exercicio 2

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)
agua = df[df["produto"] == "Água"]
vendas = agua.groupby("semana")["quantidade"].sum()
vendas.plot(kind="bar")
plt.title("Água vendida por semana")
plt.xlabel("Semana")
plt.ylabel("Quantidade")
plt.show()
print(vendas)
```

# Atividade

```python
df["receita"] = df["preco"] * df["quantidade"]
semana = df.groupby("semana")["receita"].sum()
semana.plot(kind="line", marker="o")
plt.title("Receita por semana")
plt.xlabel("Semana")
plt.ylabel("Euros")
plt.savefig("receita_semana.png")
plt.show()

categoria = df.groupby("categoria")["receita"].sum()
categoria.plot(kind="bar")
plt.title("Receita por categoria")
plt.xlabel("Categoria")
plt.ylabel("Euros")
plt.tight_layout()
plt.savefig("receita_categoria.png")
plt.show()
```
