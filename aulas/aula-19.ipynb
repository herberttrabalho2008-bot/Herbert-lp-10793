# Exercicio 1 

```python
class Veiculo:
 def __init__(self, marca):
    self.marca = marca
    self.rodas = 4
 def descrever(self):
    print(f"{self.marca} com {self.rodas} rodas")
    
class Mota(Veiculo):
 def empinar(self):
    print(f"A {self.marca} empinou!")
    
m = Mota("Yamaha")
m.descrever()
m.empinar()
# vai mostrar yamaha com 4 rodas
# vai mostrar A yamaha empinou!
```

# Exercicio 2

```python
class Forma:
    def area(self):
        return 0
class Quadrado(Forma):
    def __init__(self, lado):
        self.lado = lado
    def area(self):
        return self.lado ** 2
class Circulo(Forma):
    def __init__(self, raio):
        self.raio = raio
    def area(self):
        return 3.14 * self.raio ** 2
for f in [Quadrado(3), Circulo(2)]:
 print(f.area())
```

# Exercicio 3

```python
class Veiculo:
    def __init__(self, marca):
        self.marca = marca
        self.rodas = 4
    def descrever(self):
        print(f"{self.marca} com {self.rodas} rodas")
class Mota(Veiculo):
    def __init__(self, marca):
        super().__init__(marca)
        self.rodas = 2
        
    def empinar(self):
        print(f"A {self.marca} empinou!")
        
m = Mota("Yamaha")
m.descrever()
m.empinar()
```

# Atividade

```python
class funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario
    def pagamento(self):
        return self.salario
class gestor(funcionario):
    def __init__(self, nome, salario, bonus):
        super().__init__(nome, salario)
        self.bonus = bonus
    def pagamento(self):
        return super().pagamento() + self.bonus
class estagiario(funcionario):
    def pagamento(self):
        return super().salario / 2

equipa = [funcionario("Rui", 1000), gestor("Marta", 1500, 300), estagiario("Tiago", 800)]
total = 0
for f in equipa:
    print(f"{f.nome}: {f.pagamento():.2f} €")
    total += f.pagamento()
print(f"Total: {total:.2f} €")
```
