(3) Com que valor fica n depois do ciclo? -1
(4) O que acontecia se a linha n = n -2 não existisse? seria um ciclo infinito

```python
n = 5
while n > 0:
    print(n)
    n = n -2
print("Acabou com n =", n)
```

Exercício 2: completa os espaços

```python
# Parte A —somar de 1 a 10
total = 0
for i in range(1, 11):
    total = total + i
print("Soma de 1 a 10:", total)
# Parte B —contagem 5 4 3 2 1
for contagem in range(5, 0, -1):
    print(contagem)
print("Lançar!")
```
