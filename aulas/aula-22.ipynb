# Exercicio 2

```python
import numpy as np

p = np.array([2, 4, 6, 8])
q = np.array([1, 1, 2, 2])
print(p - q)
# vai mostrar [1 3 4 6]
print(p * q)
# vai mostrar [ 2 3 12 16]
print(p / 2)
# vai mostrar [1. 2. 3. 4.]
print(p[p > 4])
# vai mostrar [6 8]
print((p + q) > 6)
# vai mostrar [False False True True]
```

# Exercicio 3

```python
import numpy as np
temp = np.array([14, 17, 12, 19, 21, 15, 14])
media = temp.mean()
maximo = temp.max()
posicao = temp.argmax()
frios = (temp < 15).sum()
print(media, maximo, posicao, frios)
```
