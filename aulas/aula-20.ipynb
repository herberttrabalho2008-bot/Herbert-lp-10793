# Exercicio 1

```python
def saudacao(lingua):
    def dizer(nome):
        if lingua == "pt":
            return f"Olá, {nome}"
        return f"Hello, {nome}"
    return dizer

pt = saudacao("pt")
en = saudacao("en")
print(pt("Ana"))
print(en("Rui"))
```

# Exercicio 3

```python
class Temperatura:
    def __init__(self, celsius):
        self.celsius = celsius
        
    @staticmethod
    def c_para_f(c):
        return c * 9 / 5 + 32
    @classmethod
    def de_fahrenheit(cls, f):
        return cls((f - 32) * 5 / 9)
    @classmethod
    def de_kelvin(cls, k):
        return cls(k - 273.15)
    
t = Temperatura.de_kelvin(300)
print(f"{t.celsius:.2f}")
```

# Atividade

```python
class Retangulo():
    def __init__(self, largura, altura):   
        self.largura = largura
        self.altura = altura
    @property
    def area(self):
        return self.largura * self.altura
    @staticmethod
    def valido(medida):
        return medida > 0
    @classmethod
    def quadrado(cls, lado):
        return cls(lado, lado)
    
r = Retangulo(4, 3)
q = Retangulo.quadrado(5)
print(r.area)
print(q.area)
print(Retangulo.valido(-2))
```
