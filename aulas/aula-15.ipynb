# Exercício 2

```python
def pares_ate(n):
    i = 0
    while i <= n:
        yield i
        i += 2
for numero in pares_ate(10):
    print(numero)
```

# qual é mais rapido

```python
import timeit
import random
numeros = [random.randint(1, 5000) for _ in range(3000)]
 
def tem_duplicados_lista(lista):
    if numeros in lista:
        lista.append(numeros)
 
def tem_duplicados_set(lista):
    if numeros in lista:
        set()
 
print("Tempo de execução da função tem_duplicados_lista:", timeit.timeit(lambda: tem_duplicados_lista(numeros), number=20))
print("Tempo de execução da função tem_duplicados_set:", timeit.timeit(lambda: tem_duplicados_set(numeros), number=20))
#o valor não é sempre o mesmo pois o tempo de execução varia de acordo com a maquina onde esta sendo usando.
#o tempo da função tem_duplicados_set é menor do que o tempo da função tem_duplicados_lista, pois a função tem_duplicados_set utiliza um set, que é uma estrutura de dados mais eficiente para verificar se um elemento está presente ou não, enquanto a função tem_duplicados_lista utiliza uma lista, que é menos eficiente para essa operação.
```
