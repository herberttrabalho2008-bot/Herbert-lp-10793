## Exercício 1: prevê a função

## vai usar a função com os dois argumentos para fazer a area de um retangulo que é 4 e 3 para o primeiro print vai ser 12 e o segundo print vai ser 10 porque a função esta sendo executada em uma variavel para mostrar em um print a area do retangulo que vai ser 10

```python
def area_retangulo(largura, altura):
    return largura * altura
print(area_retangulo(4, 3))
resultado = area_retangulo(5, 2)
print(f"Área: {resultado} m²")
```

## Exercício 2: completa a função com desconto

```python
def preco_final(preco, iva=0.23):
    return preco + preco * iva
print(preco_final(100))
# vai ser 123.0, pois o valor do iva é 0.23 por defeito
print(preco_final(100, 0.06))
# vai ser 106.0
```

## Exercício 4: junta as duas funções

```python
def media(notas):
    return sum(notas) / len(notas)
def classifica(media_final, minima=10):
    if media_final >= minima:
        return "Aprovado"
    return "Reprovado"
turma_a = [12, 8, 15]
turma_b = [16, 18, 14]
# mostra a média e a classificação de cada turma
media_a = media(turma_a)
print(f"Média da turma A: {media_a:.2f}")
classifica_a = classifica(media_a)
print(f"Classificação da turma A: {classifica_a}")
media_b = media(turma_b)
print(f"Média da turma B: {media_b:.2f}")
classifica_b = classifica(media_b)
print(f"Classificação da turma B: {classifica_b}")

```

## Atividade: calculadora de médias

```python
def media(notas):
    return sum(notas) / len(notas)

def classifica(media_final, minima=10):
    if media_final >= minima:
        return "Aprovado"
    return "Reprovado"
#----------------
notas = []
for i in range(3):
    Valor = input(f"Nota {i+1}: ")
    Valor = Valor.replace(",", ".")
    notas.append(float(Valor))

media_final = media(notas)
classificacao = classifica(media_final)
print(f"Média: {media_final:.2f}")
print(f"Classificação: {classificacao}")
```
