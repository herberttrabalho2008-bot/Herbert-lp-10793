# Perguntas rápidas

```python
import pandas as pd

df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)
print(df["quantidade"].sum())
# vai somas as quantidades 
print(df[df["categoria"] == "Fruta"]["quantidade"].sum())
# vai somar as quantidades de frutas por semana
print(df.groupby("semana")["quantidade"].sum())
# vai mostar uma tabela com a soma das quantidades por categoria
```

# Exercício 1

```python
import pandas as pd
df = pd.read_csv("cantina.csv")
df = df.drop_duplicates()
df = df.dropna(subset=["quantidade"])
df["quantidade"] = df["quantidade"].astype(int)

sem3 = df[df["semana"] == 3]
total = sem3.groupby("categoria")["quantidade"].sum()
print(total)
```

# Exercício 2

```python
import matplotlib.pyplot as plt
# usa o df limpo do exercício 1
df["receita"] = df["preco"] * df["quantidade"]
receita = df.groupby("categoria")["quantidade"].sum()
receita.plot(kind="bar", color="steelblue")
plt.title("Receita por categoria")
plt.xlabel("Categoria")
plt.ylabel("Euros")
plt.tight_layout()
plt.show()
print(receita.round(2))
```
