# Exercico 1

```python
class Livro:
    def __init__(self, titulo, paginas):
        self.titulo = titulo
        self.paginas = paginas

l1 = Livro("Python Fácil", 300)
l2 = Livro("Contos", 100)
l2.paginas = 120

print(l1.titulo, l1.paginas)
# vai imprimir: Python Fácil 300
print(l2.titulo, l2.paginas)
# vai imprimir: Contos 120
print(l1.paginas + l2.paginas)
# vai imprimir: 420
```

# Exercicio 2

```python
class Retangulo:
    def __init__(self, largura, altura):
        self.largura = largura
        self.altura = altura
    def area(self):
        return self.largura * self.altura
r = Retangulo(4, 3)
print(r.area())
```

# Exercicio 3

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco
    def com_iva(self):
        return self.preco * 1.23
cesto = [Produto("Pão", 1.20),
         Produto("Leite", 0.90)]

for p in cesto:
    print(f"{p.nome}: {p.com_iva():.2f} €")
```

# Atividade

```python
class jogador: 
    def __init__(self,nome):
        self.nome = nome
        self.pontos = 0
    def ganhar(self, pontos):
        self.pontos += pontos
        return 
    def mostrar(self):
        print(f"{self.nome} tem {self.pontos} pontos")
j1 = jogador("Ana")
j2 = jogador("Rui")
j1.ganhar(10)
j1.ganhar(5)
j2.ganhar(7)
j1.mostrar()
j2.mostrar()

```
