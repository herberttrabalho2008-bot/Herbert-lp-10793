Exercício 2: outra função, o mesmo aplica_a_todos

```python
def quadrado(x):
 return x ** 2
def dobro(x):
 return x * 2
def aplica_a_todos(lista, funcao):
    resultado = []
    for elemento in lista:
        resultado.append(funcao(elemento))
    return resultado
numeros = [1, 2, 3, 4]
print(aplica_a_todos(numeros, quadrado))
print(aplica_a_todos(numeros, dobro))
```

Exercício 3: completa a compreensão

```python
cubos = [n ** 3 for n in range(1, 6)]
# queremos [1, 8, 27, 64, 125]
impares = [n for n in range(1, 11) if n % 2 == 1] # queremos só os ímpares
print(cubos)
#[1, 8, 27, 64, 125]
print(impares)
#[1, 3, 5, 7, 9]
```

Exercício 4: completa a compreensão de dicionário

```python
notas = {"Ana": 14, "Bruno": 9, "Carla": 18, 
"Duarte": 6}
reprovados = {nome: nota for nome, nota in notas.items() if nota < 10}
print(reprovados)
# {'Bruno': 9, 'Duarte': 6}
```
